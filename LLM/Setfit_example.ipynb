# Few-Shot Classification

Few-Shot classification: supervised classification

A classifier learn target labels based only on a **few labeled examples**.

Can be applied for a classification task with not too many labeled data points available.


**SetFit**: an efficient framework to perform few-shot classification

```python
from datasets import load_dataset
data = load_dataset('rotten_tomatoes')
train_data = data['train']
test_data = data['test']
print(str(train_data))
print(str(test_data))
```

**SetFit**

```python
from setfit import sample_dataset
sampled_train_data = sample_dataset(train_data, num_samples=16)
```

```python
from setfit import SetFitModel
model=SetFitModel.from_pretrained("sentence-transformers/paraphrase-mpnet-base-v2")
```

```python
model
```

```python
from setfit import TrainingArguments as SetFitTrainingArguments
from setfit import Trainer as SetFitTrainer
```

```python
args=SetFitTrainingArguments(
    num_epochs=1,
    num_iterations=20
)
```

```python
args.eval_strategy = args.evaluation_strategy
```

```python
trainer = SetFitTrainer(
    model=model,
    args=args,
    train_dataset=sampled_train_data,
    eval_dataset=test_data,
    metric='f1'
)
```

```python
trainer.train()
```

```python
trainer.evaluate()
```
